Ejercicio No 4

```python
class CalculadorEdades:
    def __init__(self, edad_juan: float):
        self.edad_juan = edad_juan

    def calcular_edad_alberto(self) -> float:
        return (2 / 3) * self.edad_juan

    def calcular_edad_ana(self) -> float:
        return (4 / 3) * self.edad_juan

    def calcular_edad_mama(self) -> float:
        return self.edad_juan + self.calcular_edad_alberto() + self.calcular_edad_ana()


def main():
    edad_juan = float(input("Ingrese la edad de Juan: "))
    calculador = CalculadorEdades(edad_juan)

    edad_alberto = calculador.calcular_edad_alberto()
    edad_ana = calculador.calcular_edad_ana()
    edad_mama = calculador.calcular_edad_mama()

    print(f"Edad de Juan: {edad_juan}")
    print(f"Edad de Alberto: {edad_alberto}")
    print(f"Edad de Ana: {edad_ana}")
    print(f"Edad de la mamá: {edad_mama}")


if __name__ == "__main__":
    main()
```

Ejercicio No 12

```python
class Empleado:
    def __init__(self, horas_trabajadas: float, valor_hora: float, porcentaje_retencion: float):
        self.horas_trabajadas = horas_trabajadas
        self.valor_hora = valor_hora
        self.porcentaje_retencion = porcentaje_retencion

    def calcular_salario_bruto(self) -> float:
        return self.horas_trabajadas * self.valor_hora

    def calcular_retencion_fuente(self) -> float:
        return self.calcular_salario_bruto() * (self.porcentaje_retencion / 100)

    def calcular_salario_neto(self) -> float:
        return self.calcular_salario_bruto() - self.calcular_retencion_fuente()

def main():
    horas = 48
    valor_hora = 5000.0
    porcentaje_retencion = 12.5

    empleado = Empleado(horas, valor_hora, porcentaje_retencion)

    salario_bruto = empleado.calcular_salario_bruto()
    retencion = empleado.calcular_retencion_fuente()
    salario_neto = empleado.calcular_salario_neto()

    print(f"Salario Bruto: ${salario_bruto:,.2f}")
    print(f"Retención en la fuente ({porcentaje_retencion}%): ${retencion:,.2f}")
    print(f"Salario Neto: ${salario_neto:,.2f}")

if __name__ == "__main__":
    main()
```

Ejercicio No 14

```python
class CalculadorPotencias:
    def __init__(self, numero: float):
        self.numero = numero

    def calcular_cuadrado(self) -> float:
        return self.numero ** 2

    def calcular_cubo(self) -> float:
        return self.numero ** 3

def main():
    numero = float(input("Ingrese un número: "))
    calculador = CalculadorPotencias(numero)

    cuadrado = calculador.calcular_cuadrado()
    cubo = calculador.calcular_cubo()

    print(f"Número: {numero}")
    print(f"Cuadrado: {cuadrado}")
    print(f"Cubo: {cubo}")

if __name__ == "__main__":
    main()
```

Ejercicio No 17

```python
import math

class Circulo:
    def __init__(self, radio: float):
        self.radio = radio

    def calcular_area(self) -> float:
        return math.pi * (self.radio ** 2)

    def calcular_longitud_circunferencia(self) -> float:
        return 2 * math.pi * self.radio

def main():
    radio = float(input("Ingrese el radio del círculo: "))
    circulo = Circulo(radio)

    area = circulo.calcular_area()
    longitud = circulo.calcular_longitud_circunferencia()

    print(f"Área del círculo: {area:.4f}")
    print(f"Longitud de la circunferencia: {longitud:.4f}")

if __name__ == "__main__":
    main()
```
